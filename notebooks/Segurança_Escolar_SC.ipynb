```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
import joblib

print("======================================================")
print(" TESTE 1: BASE DE DENUNCIAS FICTICIAS")
print("======================================================\n")

# 1. Carrega base de dados e trata nulos
df_sintetico = pd.read_csv('denuncias_escolares_sinteticas.csv')
df_sintetico = df_sintetico.dropna(subset=['texto_denuncia', 'nivel_risco'])
print(f"Base carregada | Num de denúncias: {len(df_sintetico)}")

X_sint = df_sintetico['texto_denuncia']
y_sint = df_sintetico['nivel_risco']

# 2. Divisão (Split) ANTES da vetorização
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_sint, y_sint, test_size=0.3, random_state=42)

# 3. Vetorização ciente do vazamento de dados
vectorizer_sint = TfidfVectorizer(max_features=5000)
# FIT apenas no treino
X_train_vet = vectorizer_sint.fit_transform(X_train_s)
# TRANSFORM apenas no teste (o modelo não "aprende" o vocabulário do teste)
X_test_vet = vectorizer_sint.transform(X_test_s)

# 4. Treinamento
lr_sint = LogisticRegression(max_iter=1000)
lr_sint.fit(X_train_vet, y_train_s)

rf_sint = RandomForestClassifier(random_state=42)
rf_sint.fit(X_train_vet, y_train_s)

# 5. Resultados Robustos
print("\n--- Resultados: Regressão Logística ---")
print(classification_report(y_test_s, lr_sint.predict(X_test_vet)))

print("\n--- Resultados: Random Forest ---")
print(classification_report(y_test_s, rf_sint.predict(X_test_vet)))

# 6. Exportação
joblib.dump(rf_sint, 'modelo_risco_escolar.joblib')
joblib.dump(vectorizer_sint, 'vetorizador_risco.joblib')
print("\nModelo Random Forest e Vetorizador salvos com sucesso!")
```

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
# Adicionamos accuracy_score e precision_score aqui:
from sklearn.metrics import classification_report, f1_score, accuracy_score, precision_score
from scipy.sparse import hstack

# ==========================================================
# CONFIGURAÇÕES
# ==========================================================
MAX_FEATURES_TOLD = 1000
MAX_FEATURES_ESCOLA = 1000
RANDOM_STATE = 42
TEST_SIZE = 0.30

# ==========================================================
# ESTÁGIO 1 — DETECÇÃO DE TOXICIDADE (ToLD-BR)
# ==========================================================
print("=" * 60)
print(" ESTÁGIO 1: DETECÇÃO DE TOXICIDADE — ToLD-BR")
print("=" * 60)

df_told = pd.read_csv("ToLD-BR.csv")
df_told["text"] = df_told["text"].fillna("")

colunas_ofensa = ["homophobia", "obscene", "insult", "racism", "misogyny", "xenophobia"]
df_told["toxicidade_geral"] = df_told[colunas_ofensa].sum(axis=1).apply(lambda x: 1 if x > 0 else 0)

X_told = df_told["text"]
y_told = df_told["toxicidade_geral"]

X_train_t, X_test_t, y_train_t, y_test_t = train_test_split(
    X_told, y_told, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y_told
)

vectorizer_told = TfidfVectorizer(max_features=MAX_FEATURES_TOLD)
X_train_tfidf = vectorizer_told.fit_transform(X_train_t)
X_test_tfidf = vectorizer_told.transform(X_test_t)

print("Treinando modelos do Estágio 1...\n")

# Modelos para competição
lr_told = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE, class_weight='balanced')
rf_told = RandomForestClassifier(random_state=RANDOM_STATE, class_weight='balanced')

lr_told.fit(X_train_tfidf, y_train_t)
rf_told.fit(X_train_tfidf, y_train_t)

# Avaliação: Calculando Acurácia, Precisão e F1-Score (Média Ponderada)
pred_lr_t = lr_told.predict(X_test_tfidf)
pred_rf_t = rf_told.predict(X_test_tfidf)

# Métricas Regressão Logística
acc_lr_t = accuracy_score(y_test_t, pred_lr_t)
prec_lr_t = precision_score(y_test_t, pred_lr_t, average='weighted', zero_division=0)
f1_lr_t = f1_score(y_test_t, pred_lr_t, average='weighted')

# Métricas Random Forest
acc_rf_t = accuracy_score(y_test_t, pred_rf_t)
prec_rf_t = precision_score(y_test_t, pred_rf_t, average='weighted', zero_division=0)
f1_rf_t = f1_score(y_test_t, pred_rf_t, average='weighted')

print("[Regressão Logística]")
print(f"  Acurácia: {acc_lr_t:.4f} | Precisão: {prec_lr_t:.4f} | F1-Score: {f1_lr_t:.4f}")
print("[Random Forest]")
print(f"  Acurácia: {acc_rf_t:.4f} | Precisão: {prec_rf_t:.4f} | F1-Score: {f1_rf_t:.4f}\n")

# Decisão automatizada pelo F1-Score
if f1_lr_t >= f1_rf_t:
    print(f"🏆 Vencedor Estágio 1: Regressão Logística (F1-Score: {f1_lr_t:.4f})")
    modelo_toxicidade = lr_told
else:
    print(f"🏆 Vencedor Estágio 1: Random Forest (F1-Score: {f1_rf_t:.4f})")
    modelo_toxicidade = rf_told

joblib.dump(modelo_toxicidade, "modelo_toxicidade_told.joblib")
joblib.dump(vectorizer_told, "vetorizador_told.joblib")
print("Modelos do Estágio 1 salvos.\n")


# ==========================================================
# ESTÁGIO 2 — CLASSIFICAÇÃO DA GRAVIDADE (DENÚNCIAS ESCOLARES)
# ==========================================================
print("=" * 60)
print(" ESTÁGIO 2: CLASSIFICAÇÃO DE GRAVIDADE")
print("=" * 60)

df_sintetico = pd.read_csv("denuncias_escolares_sinteticas.csv")
df_sintetico = df_sintetico.dropna(subset=["texto_denuncia", "nivel_risco"])

X_sint = df_sintetico["texto_denuncia"]
y_sint = df_sintetico["nivel_risco"]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_sint, y_sint, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y_sint
)

# Probabilidade do modelo vencedor do Estágio 1
prob_tox_train = modelo_toxicidade.predict_proba(vectorizer_told.transform(X_train_s))[:, 1].reshape(-1, 1)
prob_tox_test = modelo_toxicidade.predict_proba(vectorizer_told.transform(X_test_s))[:, 1].reshape(-1, 1)

# Novo TF-IDF focado no ambiente escolar
vectorizer_escola = TfidfVectorizer(max_features=MAX_FEATURES_ESCOLA)
X_train_s_tfidf = vectorizer_escola.fit_transform(X_train_s)
X_test_s_tfidf = vectorizer_escola.transform(X_test_s)

# Construção das features combinadas
X_train_pipeline = hstack([X_train_s_tfidf, prob_tox_train])
X_test_pipeline = hstack([X_test_s_tfidf, prob_tox_test])

print("Treinando modelos do Estágio 2 (Combinando Texto + Probabilidade Toxicidade)...\n")

lr_risco = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE, class_weight='balanced')
rf_risco = RandomForestClassifier(random_state=RANDOM_STATE, class_weight='balanced')

lr_risco.fit(X_train_pipeline, y_train_s)
rf_risco.fit(X_train_pipeline, y_train_s)

# Avaliação Estágio 2
pred_lr_r = lr_risco.predict(X_test_pipeline)
pred_rf_r = rf_risco.predict(X_test_pipeline)

# Métricas Regressão Logística
acc_lr_r = accuracy_score(y_test_s, pred_lr_r)
prec_lr_r = precision_score(y_test_s, pred_lr_r, average='weighted', zero_division=0)
f1_lr_r = f1_score(y_test_s, pred_lr_r, average='weighted')

# Métricas Random Forest
acc_rf_r = accuracy_score(y_test_s, pred_rf_r)
prec_rf_r = precision_score(y_test_s, pred_rf_r, average='weighted', zero_division=0)
f1_rf_r = f1_score(y_test_s, pred_rf_r, average='weighted')

print("[Regressão Logística]")
print(f"  Acurácia: {acc_lr_r:.4f} | Precisão: {prec_lr_r:.4f} | F1-Score: {f1_lr_r:.4f}")
print("[Random Forest]")
print(f"  Acurácia: {acc_rf_r:.4f} | Precisão: {prec_rf_r:.4f} | F1-Score: {f1_rf_r:.4f}\n")

# Decisão automatizada
if f1_lr_r >= f1_rf_r:
    print(f"🏆 Vencedor Estágio 2: Regressão Logística (F1-Score: {f1_lr_r:.4f})")
    modelo_risco = lr_risco
else:
    print(f"🏆 Vencedor Estágio 2: Random Forest (F1-Score: {f1_rf_r:.4f})")
    modelo_risco = rf_risco

joblib.dump(modelo_risco, "modelo_risco_escolar.joblib")
joblib.dump(vectorizer_escola, "vetorizador_escola.joblib")
print("Modelos do Estágio 2 salvos com sucesso.\n")


# ==========================================================
# TESTE COMPLETO DO PIPELINE
# ==========================================================
print("=" * 60)
print(" TESTE DO PIPELINE COM O MELHOR MODELO")
print("=" * 60)

exemplo = ["Um aluno é constantemente humilhado pelos colegas no recreio, que fazem comentários ofensivos sobre sua aparência."]

prob_tox = modelo_toxicidade.predict_proba(vectorizer_told.transform(exemplo))[:, 1][0]
print(f"Probabilidade de toxicidade (Modelo 1 Vencedor): {prob_tox:.4f}")

exemplo_pipeline = hstack([
    vectorizer_escola.transform(exemplo),
    np.array([[prob_tox]])
])

gravidade = modelo_risco.predict(exemplo_pipeline)[0]
print(f"Classificação de gravidade (Modelo 2 Vencedor): {gravidade}")
```
