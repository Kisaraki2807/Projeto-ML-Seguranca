```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import joblib

print("======================================================")
print(" TESTE 1: BASE SINTÉTICA (CLASSIFICAÇÃO DE RISCO)")
print("======================================================\n")

# 1. CARREGANDO A BASE
df_sintetico = pd.read_csv('denuncias_escolares_sintetico.csv')
# Remove possíveis valores vazios para não dar erro
df_sintetico = df_sintetico.dropna(subset=['texto_denuncia', 'nivel_risco'])
print(f"Base Sintética carregada! Total de denúncias simuladas: {len(df_sintetico)}")

# 2. PRÉ-PROCESSAMENTO (TF-IDF)
X_sint = df_sintetico['texto_denuncia']
y_sint = df_sintetico['nivel_risco']

vectorizer_sint = TfidfVectorizer(max_features=5000)
X_vetorizado_sint = vectorizer_sint.fit_transform(X_sint)

# 3. DIVISÃO E TREINAMENTO
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_vetorizado_sint, y_sint, test_size=0.3, random_state=42)

lr_sint = LogisticRegression(max_iter=1000)
lr_sint.fit(X_train_s, y_train_s)

rf_sint = RandomForestClassifier(random_state=42)
rf_sint.fit(X_train_s, y_train_s)

# 4. RESULTADOS E EXPORTAÇÃO
print("\n--- Resultados (Base Sintética) ---")
print(f"Acurácia - Regressão Logística: {accuracy_score(y_test_s, lr_sint.predict(X_test_s)) * 100:.2f}%")
print(f"Acurácia - Random Forest: {accuracy_score(y_test_s, rf_sint.predict(X_test_s)) * 100:.2f}%")

joblib.dump(rf_sint, 'modelo_risco_escolar.joblib')
joblib.dump(vectorizer_sint, 'vetorizador_risco.joblib')
print("\n[SUCESSO] Modelos da Base Sintética exportados (.joblib)!")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import joblib

print("======================================================")
print(" TESTE 2: BASE ToLD-BR (CLASSIFICAÇÃO DE TOXICIDADE)")
print("======================================================\n")

# 1. CARREGANDO A BASE
df_told = pd.read_csv('ToLD-BR.csv')
print(f"Base ToLD-BR carregada! Total de textos: {len(df_told)}")

# 2. PRÉ-PROCESSAMENTO E ENGENHARIA DE ATRIBUTOS
# Preenche valores vazios no texto com espaço em branco
df_told['text'] = df_told['text'].fillna('')

# Cria uma coluna alvo simplificada: Se tiver qualquer ofensa (>0), é Tóxico (1), senão é Limpo (0)
colunas_ofensa = ['homophobia', 'obscene', 'insult', 'racism', 'misogyny', 'xenophobia']
df_told['toxicidade_geral'] = df_told[colunas_ofensa].sum(axis=1).apply(lambda x: 1 if x > 0 else 0)

X_told = df_told['text']
y_told = df_told['toxicidade_geral']

vectorizer_told = TfidfVectorizer(max_features=5000)
X_vetorizado_told = vectorizer_told.fit_transform(X_told)

# 3. DIVISÃO E TREINAMENTO
X_train_t, X_test_t, y_train_t, y_test_t = train_test_split(X_vetorizado_told, y_told, test_size=0.3, random_state=42)

lr_told = LogisticRegression(max_iter=1000)
lr_told.fit(X_train_t, y_train_t)

rf_told = RandomForestClassifier(random_state=42)
rf_told.fit(X_train_t, y_train_t)

# 4. RESULTADOS E EXPORTAÇÃO
print("\n--- Resultados (Base ToLD-BR) ---")
print(f"Acurácia - Regressão Logística: {accuracy_score(y_test_t, lr_told.predict(X_test_t)) * 100:.2f}%")
print(f"Acurácia - Random Forest: {accuracy_score(y_test_t, rf_told.predict(X_test_t)) * 100:.2f}%")

joblib.dump(rf_told, 'modelo_toxicidade_told.joblib')
joblib.dump(vectorizer_told, 'vetorizador_told.joblib')
print("\n[SUCESSO] Modelos da Base ToLD-BR exportados (.joblib)!")
```
